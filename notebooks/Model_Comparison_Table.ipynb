```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python

results = pd.DataFrame({
    "Model": [
        "XGBoost",
        "Random Forest",
        "Logistic Regression",
        "SVM"
    ],
    "Accuracy (%)": [
        85.25,
        83.61,
        80.33,
        78.69
    ]
})

print(results)

results.plot(
    x="Model",
    y="Accuracy (%)",
    kind="bar",
    legend=False,
    ylim=(0, 100),
    rot=20
)

plt.title("Heart Disease Prediction - Model Comparison")
plt.ylabel("Accuracy (%)")
plt.tight_layout()
plt.show()
```
